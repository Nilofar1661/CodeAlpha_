```python
summary = pd.DataFrame({
    'metrics' : [
       'Total sales',
       'Total profit',
       'Total quantity',
       'Average discount'
    ],


    'Values': [
        int(df['Sales'].sum()),
        int(df['Profit'].sum()),
        int(df['Quantity'].sum()),
        int(df['Discount'].mean()*100)
    ]

})
print(summary)
```

```python
category_analysis = df.groupby('Category').agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum'),
    Quantity=('Quantity', 'sum')
).sort_values('Sales', ascending=False)

category_analysis
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8, 5))

sns.barplot(
    data=category_analysis.reset_index(),
    x='Category',
    y='Sales'
)

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=category_analysis.reset_index(),
    x='Category',
    y='Profit'
)

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.show()
```

```python
subcategory_analysis = df.groupby('Sub-Category').agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum'),
    Quantity=('Quantity', 'sum')
).sort_values('Sales', ascending=False)

subcategory_analysis
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=subcategory_analysis.reset_index(),
    x='Sub-Category',
    y='Sales'
)

plt.title("Sales by Sub-Category")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=subcategory_analysis.reset_index(),
    x='Sub-Category',
    y='Profit'
)

plt.title("Profit by Sub-Category")
plt.xticks(rotation=45)

plt.show()
```

```python
top_products_sales = (
    df.groupby('Product Name')['Sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products_sales
```

```python
top_products_profit = (
    df.groupby('Product Name')['Profit']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products_profit
```

```python
plt.figure(figsize=(12, 6))

top_products_profit.sort_values().plot(
    kind='barh'
)

plt.title("Top 10 Products by Profit")
plt.xlabel("Total Profit")
plt.ylabel("Product")

plt.show()
```

```python
bottom_products_profit = (
    df.groupby('Product Name')['Profit']
    .sum()
    .sort_values()
    .head(10)
)

bottom_products_profit
```

```python
segment_analysis = df.groupby('Segment').agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum'),
    Quantity=('Quantity', 'sum')
)

segment_analysis
```

```python
region_analysis = df.groupby('Region').agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum'),
    Quantity=('Quantity', 'sum')
).sort_values('Sales', ascending=False)

region_analysis
```

```python
region_analysis[['Sales', 'Profit']].plot(
    kind='bar',
    figsize=(10, 6)
)

plt.title("Sales and Profit by Region")
plt.xlabel("Region")
plt.ylabel("Amount")

plt.xticks(rotation=0)
plt.show()
```

```python
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Month Name'] = df['Order Date'].dt.month_name()
df['Quarter'] = df['Order Date'].dt.quarter
```

```python
yearly_sales = df.groupby('Year').agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum')
)

yearly_sales
```

```python
plt.figure(figsize=(10, 6))

sns.lineplot(
    data=yearly_sales,
    x=yearly_sales.index,
    y='Sales',
    marker='o'
)

plt.title("Yearly Sales Trend")
plt.xlabel("Year")
plt.ylabel("Sales")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.lineplot(
    data=yearly_sales,
    x=yearly_sales.index,
    y='Profit',
    marker='o'
)

plt.title("Yearly Profit Trend")
plt.xlabel("Year")
plt.ylabel("Profit")

plt.show()
```

```python
monthly_sales = (
    df.groupby('Month')['Sales']
    .sum()
)

plt.figure(figsize=(10, 6))

sns.lineplot(
    x=monthly_sales.index,
    y=monthly_sales.values,
    marker='o'
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xticks(range(1, 13))

plt.show()
```

```python
year_month_sales = df.pivot_table(
    index='Month',
    columns='Year',
    values='Sales',
    aggfunc='sum'
)

plt.figure(figsize=(12, 6))

sns.heatmap(
    year_month_sales,
    annot=True,
    fmt='.0f'
)

plt.title("Sales Heatmap: Year vs Month")
plt.xlabel("Year")
plt.ylabel("Month")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Discount',
    y='Profit',
    hue='Category',
    alpha=0.6
)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.show()
```

```python
category_analysis['Profit Margin (%)'] = (
    category_analysis['Profit'] /
    category_analysis['Sales']
) * 100

category_analysis
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=category_analysis.reset_index(),
    x='Category',
    y='Profit Margin (%)'
)

plt.title("Profit Margin by Category")
plt.xlabel("Category")
plt.ylabel("Profit Margin (%)")

plt.show()
```

```python
negative_profit = df[df['Profit'] < 0]

print("Number of loss-making transactions:", len(negative_profit))

negative_profit.head()
```

```python
negative_profit_percentage = (
    len(negative_profit) / len(df)
) * 100

print(
    f"Percentage of loss-making transactions: "
    f"{negative_profit_percentage:.2f}%"
)
```

```python
high_discount_losses = df[
    (df['Discount'] >= 0.3) &
    (df['Profit'] < 0)
]

print(
    "High-discount loss-making transactions:",
    len(high_discount_losses)
)

high_discount_losses.head()

df.groupby('Discount')['Profit'].mean()
```
